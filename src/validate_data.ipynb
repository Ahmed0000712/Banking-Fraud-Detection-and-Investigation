```python
import pandas as pd
from pathlib import Path

ROOT = Path.cwd()
if not (ROOT / "data").exists():
    ROOT = ROOT.parent

DATA = ROOT / "data"

customers = pd.read_csv(DATA / "customers.csv")
accounts = pd.read_csv(DATA / "accounts.csv")
transactions = pd.read_csv(DATA / "transactions.csv")
answers = pd.read_csv(DATA / "answers.csv")

print("All four CSV files loaded successfully.")
```

```python
# Check missing values in required columns
required = {
    "customers": (customers, ["CustomerID", "CustomerName"]),
    "accounts": (accounts, ["AccountID", "CustomerID"]),
    "transactions": (transactions, [
        "TransactionID", "CustomerID", "AccountID",
        "TransactionDate", "AmountINR", "Country"
    ]),
    "answers": (answers, ["TransactionID", "IsFraud", "FraudScenario"])
}

for name, (df, columns) in required.items():
    print(name, "- Missing values:", df[columns].isna().sum().sum())

# Check duplicate IDs
print("\nDuplicate IDs:")
print("Customers:", customers["CustomerID"].duplicated().sum())
print("Accounts:", accounts["AccountID"].duplicated().sum())
print("Transactions:", transactions["TransactionID"].duplicated().sum())
print("Answers:", answers["TransactionID"].duplicated().sum())
```

```python
print("Account IDs match Customer IDs:",
      (accounts["AccountID"] == accounts["CustomerID"]).all())

print("Transaction IDs match Customer IDs:",
      (transactions["AccountID"] == transactions["CustomerID"]).all())

print("All transactions are domestic:",
      (transactions["Country"] == "India").all())

print("All transaction amounts are positive:",
      (transactions["AmountINR"] > 0).all())

print("Transaction and answer IDs match:",
      set(transactions["TransactionID"]) == set(answers["TransactionID"]))
```

```python
fraud = answers[answers["IsFraud"] == 1]
legitimate = answers[answers["IsFraud"] == 0]

print("Total customers:", len(customers))
print("Total accounts:", len(accounts))
print("Total transactions:", len(transactions))
print("Fraud transactions:", len(fraud))
print("Legitimate transactions:", len(legitimate))
print("Fraud percentage:", answers["IsFraud"].mean() * 100)

print("\nFraud scenarios:")
print(fraud["FraudScenario"].value_counts())

print("\nFraud cases:")
print(fraud["FraudCaseID"].value_counts())
```

```python
checks = [
    len(customers) == 100,
    len(accounts) == 100,
    len(transactions) == 10000,
    len(fraud) == 100,
    len(legitimate) == 9900,
    customers["CustomerName"].nunique() == 100,
    (accounts["AccountID"] == accounts["CustomerID"]).all(),
    (transactions["AccountID"] == transactions["CustomerID"]).all(),
    set(transactions["TransactionID"]) == set(answers["TransactionID"])
]

print("Passed:", sum(checks), "/", len(checks))

if all(checks):
    print("MAIN VALIDATION CHECKS PASSED")
else:
    print("CHECK FAILED — REVIEW THE OUTPUT ABOVE")
```
